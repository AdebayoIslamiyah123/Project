```python
import numpy as np
import pandas as pd
```

```python
print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

```python
# Create a one-dimensional.array
sales = np.array([12000, 18000, 15000, 22000, 19500])

print(sales)
print("Type:", type(sales))
print("Dimensions:", sales.ndim)
print("shape:", sales.shape) 
print("Number of elements:", sales.size)
print("Data type:", sales.dtype)
```

```python
Weekly_sales = np.array([
    [1200, 1500, 2000],
    [1300, 1600, 1900]
])
print (Weekly_sales)
print("Dimensions:", Weekly_sales.ndim)
print("Shape:", Weekly_sales.shape)
```

```python
prices = np.array([1200, 1500, 2000, 1300, 1600, 900])

print ("add 200:", prices + 200)
print ("Double:", prices * 2)
print ("Half:", prices / 2)
print ("Sum:", sum(prices))
print ("Max:", max(prices))
print ("Min:", min(prices))
print ("Mean:", np.mean (prices))
```

```python
prices = np.array([1200, 1500, 2000, 1300, 1600, 900])
discount_rate = 0.10

discounted_prices = prices * (1-discount_rate)
print(discounted_prices)
```

```python
# Series
product_sales = pd.Series(
    [12000, 18000, 15000],
    index = ["Rice", "Beans", "Oil"],
    name="Revenue"
)
print(product_sales)
```

```python
# DataFrame
sample = pd.DataFrame({
    "Product": ["Rice", "Beans", "Oil", "Yam"],
    "Price": [52000, 38000, 27000, 41000],
    "Quantity": [10, 15, 20, 8]
})
display(sample)
```
